````python
def standardize(column):
    """Standardize the values in a column.

    Args:
        column (pandas Series): The data to standardize.

    Returns:
        pandas Series: the values as z-scores
    """
    return (column - column.mean()) / column.std()
````

### Exercise 2(e) (5 points)

Another student has written this function to calculate the mean and median of a list. You want to show them how to split it into two simpler functions: `mean()` and `median()`.

```
def mean_and_median(values):
  """Get the mean and median of a list of 'values'

  Args:
    values (iterable of float): A list of numbers

  Returns:
    tuple (float, float): The mean and median
  """
  mean = sum(values) / len(values)
  midpoint = int(len(values) / 2)
  if len(values) % 2 == 0:
    median = (values[midpoint - 1] + values[midpoint]) / 2
  else:
    median = values[midpoint]

  return mean, median
```

Split the above function into two functions called `mean()` and `median()`.

````python
def mean(values):
    """Get the mean of a list of values."""
    return sum(values) / len(values)


def median(values):
    """Get the median of a list of values."""
    midpoint = int(len(values) / 2)

    if len(values) % 2 == 0:
        return (values[midpoint - 1] + values[midpoint]) / 2
    else:
        return values[midpoint]
````

### Exercise 4(a) (6 points)

Below is a single, non-modular block of code that loads the `food` dataset, removes duplicate rows, and counts the missing values in each column:

```

```

Refactor this code into three modular functions: `load_data(path)`, `remove_duplicates(df)`, and `count_missing(df)`, where each function does exactly one task. Then call the three functions in sequence, passing the output of one as the input to the next, to reproduce the same result.

````python
def load_data(path):
    return pd.read_csv(path, low_memory=False)

def remove_duplicates(df):
    return df.drop_duplicates()

def count_missing(df):
    return df.isnull().sum()

data = load_data("openfoodfacts_org_products_sample.csv")
data = remove_duplicates(data)
missing = count_missing(data)

print(missing)
````
