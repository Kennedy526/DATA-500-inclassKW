# Exercise 4 (10 points)

A pastry shop produces two types of pastries: Croissants (X) and Danishes (Y). Each pastry requires:

| Resource | Croissant (X) | Danish (Y) | Available | 
| --- | --- | --- | --- |
| Flour (grams) | 200 | 100 | 5000 |
| Labor (minutes) | 10 | 15 | 400 |
| Profit per unit | $2 | $3 | |

The owner wants to maximize profit while not exceeding available resources. Formulate and solve this as a linear programming problem using Python.

```python
from scipy.optimize import linprog

# profit = 2X + 3Y
c = [-2, -3]

# Coefficients for the constraints
A = [
    [200, 100],
    [10, 15]
]

# Right side constraints
b = [5000, 400]

# Bounds for each variable
x_bounds = (0, None)
y_bounds = (0, None)

# Solve the linear programming problem
res = linprog(c, A_ub=A, b_ub=b, bounds=[x_bounds, y_bounds])

# Print the results
if res.success:
    print(f"Optimal profit: {-res.fun}")
    print(f"Units of X: {res.x[0]}")
    print(f"Units of Y: {res.x[1]}")
else:
    print("No solution found.")
```

# Exercise 5 (10 points)

Consider the context presented in Exercise 4. This time solve the program as an interger linear programming. Formulate and solve the problem as an integer programming problem in Python.

```python
import numpy as np
from scipy.optimize import milp, LinearConstraint, Bounds

# profit = 2X + 3Y
c = [-2.000001, -3]
# Coefficients for the constraints
A = np.array([
    [200, 100],
    [10, 15]
])

# Right side constraint
b = np.array([5000, 400])

# Setting the constraints
linear_constraint = LinearConstraint(A, lb=-np.inf, ub=b)

# Variable bounds
bounds = Bounds([0, 0], [np.inf, np.inf])

# Integer variables (X and Y must be integers)
integrality = np.ones(2, dtype=bool)

# Solve the mixed-integer linear programming problem
res = milp(
    c=c,
    constraints=[linear_constraint],
    bounds=bounds,
    integrality=integrality
)

# Print the results
if res.success:
    print(f"Optimal profit: {round(-res.fun)}")
    print(f"Units of X: {res.x[0]}")
    print(f"Units of Y: {res.x[1]}")
else:
    print("No solution found.")
```

```python
import numpy as np
from scipy.optimize import milp, LinearConstraint, Bounds

# profit = 2X + 3Y
c = [-2, -3]
# Coefficients for the constraints
A = np.array([
    [200, 100],
    [10, 15]
])

# Right side constraint
b = np.array([5000, 400])

# Setting the constraints
linear_constraint = LinearConstraint(A, lb=-np.inf, ub=b)

# Variable bounds
bounds = Bounds([0, 0], [np.inf, np.inf])

# Integer variables (X and Y must be integers)
integrality = np.ones(2, dtype=bool)

# Solve the mixed-integer linear programming problem
res = milp(
    c=c,
    constraints=[linear_constraint],
    bounds=bounds,
    integrality=integrality
)

# Print the results
if res.success:
    print(f"Optimal profit: {round(-res.fun)}")
    print(f"Units of X: {res.x[0]}")
    print(f"Units of Y: {res.x[1]}")
else:
    print("No solution found.")
```
